```python
import pandas as pd

url = "https://raw.githubusercontent.com/rolandmueller/titanic/main/titanic3.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

**## 7. Save Clean Dataset**

After cleaning the dataset, the cleaned data is saved as a new CSV file.

The original dataset is kept unchanged, while the cleaned dataset is exported separately for further use.

```python
# Create a copy of the original dataset
clean_df = df.copy()

# Fill missing Age values with median
clean_df["age"] = clean_df["age"].fillna(clean_df["age"].median())

# Fill missing Fare values with median
clean_df["fare"] = clean_df["fare"].fillna(clean_df["fare"].median())

# Fill missing Embarked values with mode
clean_df["embarked"] = clean_df["embarked"].fillna(
    clean_df["embarked"].mode()[0]
)

print("Data cleaning completed!")
```

```python
# Save cleaned dataset
output_file = "titanic_cleaned.csv"

clean_df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("Output file:", output_file)
```

**## 9. Python Data Cleaning CLI Tool**

The data cleaning process is converted into a command-line Python script.

The CLI tool accepts an input CSV file and an output file path. It loads the input dataset, handles selected missing values, and saves the cleaned dataset.

```python
%%writefile data_cleaner.py

import argparse
import pandas as pd


def clean_dataset(input_file, output_file):
    # Load dataset
    df = pd.read_csv(input_file)

    print("Dataset loaded successfully!")
    print("Original shape:", df.shape)

    # Create a copy
    clean_df = df.copy()

    # Handle missing numerical values
    if "age" in clean_df.columns:
        clean_df["age"] = clean_df["age"].fillna(clean_df["age"].median())

    if "fare" in clean_df.columns:
        clean_df["fare"] = clean_df["fare"].fillna(clean_df["fare"].median())

    # Handle missing categorical value
    if "embarked" in clean_df.columns:
        if not clean_df["embarked"].mode().empty:
            clean_df["embarked"] = clean_df["embarked"].fillna(
                clean_df["embarked"].mode()[0]
            )

    # Save cleaned dataset
    clean_df.to_csv(output_file, index=False)

    print("Cleaning completed successfully!")
    print("Cleaned shape:", clean_df.shape)
    print("Output file:", output_file)


if __name__ == "__main__":
    parser = argparse.ArgumentParser(
        description="Python Data Cleaning CLI Tool"
    )

    parser.add_argument(
        "--input",
        required=True,
        help="Path to input CSV file"
    )

    parser.add_argument(
        "--output",
        required=True,
        help="Path to output cleaned CSV file"
    )

    args = parser.parse_args()

    clean_dataset(args.input, args.output)
```
